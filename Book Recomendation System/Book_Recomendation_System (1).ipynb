```python
ratings_with_name = ratings.merge(books, on='ISBN')
```

```python
#numbers of ratings per book
num_rating_df = ratings_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'num_ratings'},inplace=True)
```

```python
ratings['Book-Rating'] = ratings['Book-Rating'].astype(int)
```

```python
avg_rating_df = ratings_with_name.groupby('Book-Title')['Book-Rating'].mean().reset_index()
```

```python
#done with books
#now keep only those users who have votes 200 times
# votes > 200

#Count ratings per users
user_counts = ratings_with_name.groupby('User-ID').count()['Book-Rating']
```

```python
user_counts
```

```python
active_users = user_counts[user_counts > 200].index
```

```python
active_users
```

```python
filtered_rating = ratings_with_name[ratings_with_name['User-ID'].isin(active_users)]
```

```python
filtered_rating = filtered_rating.drop(['Image-URL-S', 'Image-URL-L'], axis=1)
```

```python
filtered_rating
```

```python
#filter popular books
book_counts = filtered_rating.groupby('Book-Title').count()['Book-Rating']
```

```python
book_counts
```

```python
famous_books = book_counts[book_counts > 50].index
```

```python
famous_books #popular
```

```python
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
final_ratings
```

```python
#create pivot table
pt = final_ratings.pivot_table(
    index='Book-Title',
    columns='User-ID',
    values='Book-Rating',
)
pt.fillna(0,inplace = True)
```

```python
pt
```

```python
pt.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(pt)
```

```python
def recommend(book_name):
    if book_name not in pt.index:
        print("Book not found")
        return

    index = np.where(pt.index == book_name)[0][0]

    similar_items = sorted(
        list(enumerate(similarity_scores[index])),
        key=lambda x: x[1],
        reverse=True
    )[1:6]

    for i in similar_items:
        print(pt.index[i[0]])
```
